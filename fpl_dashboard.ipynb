```python
import numpy as np

players["xgc_per90"] = players["expected_goals_conceded"] / players["nineties"]

active = players[players["minutes"] >= 300]
for col, name in [("tp_per90", "pos_median_tp90"), ("xgi_per90", "pos_median_xgi90"),
                  ("defcon_per90", "pos_median_dc90"), ("xgc_per90", "pos_median_xgc90")]:
    players[name] = players["position"].map(active.groupby("position")[col].median())

low_pts = players["tp_per90"] < 0.8 * players["pos_median_tp90"]
is_att = players["position"].isin(["Midfielder", "Forward"])
is_def = players["position"] == "Defender"

# attackers: one process signal (underlying involvement is low)
att_poor = players["xgi_per90"] < 0.8 * players["pos_median_xgi90"]

# defenders: count weak signals (team concedes lots of chances, he contributes little)
def_signals = (players["xgc_per90"] > 1.2 * players["pos_median_xgc90"]).astype(int) \
            + (players["defcon_per90"] < 0.8 * players["pos_median_dc90"]).astype(int)

players["diagnosis"] = np.select(
    [
        ~(is_att | is_def),
        low_pts & is_att & att_poor,
        low_pts & is_att & ~att_poor,
        low_pts & is_def & (def_signals == 2),
        low_pts & is_def & (def_signals == 1),
        low_pts & is_def & (def_signals == 0),
    ],
    ["Not assessed", "Trap", "Unlucky", "Trap", "Watch", "Unlucky"],
    default="OK",
)

trap_mask = (players["minutes"] >= 300) & (players["selected_by_percent"] > 10)
cols = ["web_name", "team_name_short", "position", "selected_by_percent",
        "tp_per90", "xgi_per90", "xgc_per90", "defcon_per90", "diagnosis"]
players[trap_mask].sort_values("selected_by_percent", ascending=False)[cols]
```

```python
import numpy as np
import pandas as pd
import ipywidgets as widgets
from IPython.display import display

cols = ["web_name", "team_name", "position", "price", "total_points", "points_per_million",
        "selected_by_percent", "minutes", "tp_per90", "xgi_per90", "xgc_per90",
        "defcon_per90", "diagnosis", "news"]
player_table = players[cols].copy()

teams = ["All"] + sorted(player_table["team_name"].unique())
round_cols = {"points_per_million": 2, "tp_per90": 2, "xgi_per90": 2,
              "xgc_per90": 2, "defcon_per90": 2}

def show_player_table(position, team, sort_by, max_price, min_minutes):
    mask = (player_table["price"] <= max_price) & (player_table["minutes"] >= min_minutes)
    if position != "All":
        mask &= player_table["position"] == position
    if team != "All":
        mask &= player_table["team_name"] == team

    df = player_table[mask]
    lower_is_better = sort_by in ["xgc_per90", "price"]
    df = df.sort_values(sort_by, ascending=lower_is_better).head(15)
    display(df.round(round_cols))

widgets.interact(
    show_player_table,
    position=["All", "Goalkeeper", "Defender", "Midfielder", "Forward"],
    team=teams,
    sort_by=["total_points", "points_per_million", "selected_by_percent", "price",
             "tp_per90", "xgi_per90", "xgc_per90", "defcon_per90"],
    max_price=widgets.FloatSlider(min=4.0, max=16.0, step=0.5, value=16.0),
    min_minutes=widgets.IntSlider(min=0, max=900, step=30, value=300),
)
```

# Gameweek Summary

```python
def get_live(gw):
  url = f"https://fantasy.premierleague.com/api/event/{gw}/live/"
  gw_data = json.loads(requests.get(url).text)["elements"]
  live = pd.DataFrame(gw_data)
  stats = pd.json_normalize(live["stats"])
  out = pd.concat([live[["id"]], stats], axis=1)
  return out.merge(players[["id", "web_name", "team_name", "position"]], on="id", how="left")

def show_gameweek_summary(gw):
  row = fin_summary[fin_summary["id"] == gw].iloc[0]
  live = get_live(gw)
  print(f"GW{gw}: average {row['average_entry_score']}, highest {row['highest_score'].astype("int32")}, "
        f"captain {row['most_captained']} ({live[live["web_name"] == row["most_captained"]]["total_points"].iloc[0]} pts), top scorer {row['top_element']} ({row['top_points']} pts), "
        f"most transferred {row['most_transferred_in']}, {row["transfers_made"]} total transfers this week, "
        f"most selected {row['most_selected']}"
        )
  cols = ["web_name", "team_name", "position", "minutes", "goals_scored",
          "assists", "clean_sheets", "bonus", "total_points"]
  display(live.sort_values("total_points", ascending=False)[cols].head(10))

widgets.interact(show_gameweek_summary, gw=widgets.Dropdown(options=list(fin_summary["id"]), value=fin_summary["id"].max()))
```

# Per player details

```python
HIST_KEEP = [
    "round", "opponent_team", "was_home", "kickoff_time", "team_h_score", "team_a_score",
    "minutes", "starts",
    "total_points", "goals_scored", "assists", "clean_sheets", "goals_conceded", "saves", "bonus", "bps",
    "expected_goals", "expected_assists", "expected_goal_involvements", "expected_goals_conceded",
    "defensive_contribution", "clearances_blocks_interceptions", "tackles", "recoveries",
    "value", "selected", "transfers_balance",
]

def to_local(series):
    return pd.to_datetime(series, utc=True).dt.tz_convert("Asia/Kolkata")

def get_player(pid):
    data = requests.get(f"https://fantasy.premierleague.com/api/element-summary/{pid}/").json()
    hist = pd.DataFrame(data["history"])
    if not hist.empty:
        hist = hist[HIST_KEEP].copy()
        hist["price"] = hist["value"] / 10
        for col in ["expected_goals", "expected_assists",
                    "expected_goal_involvements", "expected_goals_conceded"]:
            hist[col] = pd.to_numeric(hist[col])
        hist["opponent_team"] = hist["opponent_team"].map(id_to_name)
        kick = to_local(hist["kickoff_time"])
        hist["kickoff"] = kick
        hist["kickoff_text"] = kick.dt.strftime("%a %d %b, %H:%M")
        hist["venue"] = np.where(hist["was_home"], "H", "A")

    fixtures = pd.DataFrame(data["fixtures"])
    if not fixtures.empty:
        # the opponent is the away team when he is at home, otherwise the home team
        opp_id = fixtures["team_a"].where(fixtures["is_home"], fixtures["team_h"])
        fixtures["opponent"] = opp_id.map(id_to_name)
        fixtures["venue"] = np.where(fixtures["is_home"], "H", "A")
        kick = to_local(fixtures["kickoff_time"])
        fixtures["kickoff_text"] = kick.dt.strftime("%a %d %b, %H:%M").fillna("TBC")
        fixtures = (fixtures[["event_name", "kickoff_text", "opponent", "venue", "difficulty"]]
                    .rename(columns={"event_name": "gameweek"}))

    return hist, fixtures


EXTRA_COLS = {
    "Forward":    ["expected_goals", "expected_assists", "expected_goal_involvements"],
    "Midfielder": ["expected_goals", "expected_assists", "expected_goal_involvements"],
    "Defender":   ["expected_goals_conceded", "defensive_contribution",
                   "clearances_blocks_interceptions", "tackles", "recoveries"],
    "Goalkeeper": ["saves", "expected_goals_conceded"],
}

def show_player(pid):
    p = players.loc[players["id"] == pid].iloc[0]
    print(f"{p['web_name']} | {p['team_name']} | {p['position']} | £{p['price']}m | "
          f"{p['selected_by_percent']}% owned | {p['total_points']} pts | status: {p['status']}")
    if p["news"]:
        print("News:", p["news"])

    hist, fixtures = get_player(pid)
    if hist.empty:
        print("No matches played yet.")
    else:
        base = ["round", "kickoff_text", "opponent_team", "venue", "minutes",
                "total_points", "goals_scored", "assists", "clean_sheets", "bonus", "price"]
        display(hist[base + EXTRA_COLS[p["position"]]].round(2))

        # sum over double gameweeks so each round is one point on the line
        per_round = hist.groupby("round", as_index=False)["total_points"].sum()
        fig = px.line(per_round, x="round", y="total_points", markers=True,
                      title=f"{p['web_name']}: points per gameweek",
                      labels={"round": "Gameweek", "total_points": "Points"})
        fig.show()

    print("Upcoming fixtures")
    if fixtures.empty:
        print("None scheduled.")
    else:
        display(fixtures.style.background_gradient(
            cmap="RdYlGn_r", subset=["difficulty"], vmin=1, vmax=5))

options = [(f"{n} ({t})", i) for n, t, i in
           sorted(zip(players["web_name"], players["team_name_short"], players["id"]))]

widgets.interact(show_player, pid=widgets.Dropdown(options=options, description="Player"))
```
